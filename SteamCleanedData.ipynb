```python
mask=df["Website"].notna() & ~df["Website"].str.contains(r'\.',na=False)
df.loc[mask, "Website"]=None
mask_kiem=df["Website"].notna() & ~df["Website"].str.contains(r'\.',na=False)
print(mask_kiem.sum())
```

```python
df=df.drop(columns={"Movies","Score rank","Metacritic url"})
```

```python
import numpy as np
for col in ["Full audio languages","Supported languages"]:

    df[col] = df[col].apply(
    lambda x: ", ".join(x) if isinstance(x, list)
    else (x if pd.isna(x) else str(x).replace('[', '').replace(']', '').replace("'","").strip())
)
df=df.replace('',np.nan)
```

```python
def save(df,unique_limit=255):
   for f in df.columns:
      if df[f].dtype== "bool":
        continue
      if df[f].dtype== "int64" and df[f].min() >=0 and df[f].max() <= 255:
         df[f]=df[f].astype("uint8")
      elif df[f].nunique() <= unique_limit:df[f]=df[f].astype("category")
   return df
df=save(df)
```

```python
print(f"Shape: {df.shape} | Null: {df.isnull().sum().sum()} | Dupe: {df.duplicated().sum()}")
print(f"\n datatypes:\n{df.dtypes.value_counts()}")
```

```python
chunk_size = 25000
base_name = "Steam_Vip_Optimized"
output_dir = '/content/drive/MyDrive/Processed_Data'
os.makedirs(output_dir, exist_ok=True)

for i in range(0, len(df), chunk_size):
    chunk_df = df.iloc[i:i + chunk_size]
    part_num = (i // chunk_size) + 1
    file_out = os.path.join(output_dir, f"{base_name}_part_{part_num}.csv")
    chunk_df.to_csv(file_out, index=False)
    print(f"📦 Created: {os.path.basename(file_out)} ({os.path.getsize(file_out) / (1024 * 1024):.2f} MB)")
```
